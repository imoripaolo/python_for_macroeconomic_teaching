```python
"""

Solow-Swann Model / Modelo Solow-Swann

Author / Autor: imoripaolo

"""

import numpy as np
import matplotlib.pyplot as plt

size = 100

n, delta, alpha, s = 0.2, 0.1, 0.3, 0.5 
```

```python
"""

Capital per Labor / Capital por Trabalhador

Version with g = 0 / Versão com g = 0

y = k**alpha
k_dot = sy - (n + delta) k = sk**alpha - (n + delta) k

"""

k = np.linspace(0,3, size)
y = np.zeros(size)
i = np.zeros(size)
dep = np.zeros(size)
k_dot = np.zeros(size)

for j in range(0,size):
    y[j] = k[j]**alpha
    i[j] = s*(k[j]**alpha)
    dep[j] = (n + delta)*k[j]
    k_dot[j] = i[j] - dep[j]

ss = np.argmin(k_dot[1:size]**2)
steady = (k[ss+1], k[ss+1]**alpha)

fig, ax = plt.subplots()
ax.plot(k, y,'b', label=r'$y=k^\alpha$')
ax.plot(k, i,'g', label=r'$sy=sk^\alpha$')
ax.plot(k, dep,'r', label=r'$(n+\delta)k$')
plt.axvline(x=steady[0], color='black', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
ax.set_ylabel('')
ax.set_xlabel('k')
plt.legend()
plt.show()


fig, ax = plt.subplots()
ax.plot(k, k_dot,'b', label=r'$\dot{k}$')
plt.axvline(x=steady[0], color='black', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
plt.axhline(y=0, color='black')
ax.set_ylabel('')
ax.set_xlabel('k')
plt.legend()
plt.show()

print(r'$k^{ss}$=', steady[0], "e $y^{ss}$=", steady[1] )
```

```python
"""

Comparative Statics / Estática Comparativa

"""

n_H, delta_H, alpha_H, s_H = 0.25, 0.2, 0.4, 0.6 
n_L, delta_L, alpha_L, s_L = 0.15, 0.05, 0.2, 0.4     

y_n_H, y_delta_H, y_alpha_H, y_s_H = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
y_n_L, y_delta_L, y_alpha_L, y_s_L = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
i_n_H, i_delta_H, i_alpha_H, i_s_H = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
i_n_L, i_delta_L, i_alpha_L, i_s_L = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
dep_n_H, dep_delta_H, dep_alpha_H, dep_s_H = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
dep_n_L, dep_delta_L, dep_alpha_L, dep_s_L = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
k_dot_n_H, k_dot_delta_H, k_dot_alpha_H, k_dot_s_H = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)
k_dot_n_L, k_dot_delta_L, k_dot_alpha_L, k_dot_s_L = np.zeros(size), np.zeros(size), np.zeros(size), np.zeros(size)

for j in range(0,size):
    y_n_H[j], y_delta_H[j], y_alpha_H[j], y_s_H[j] = k[j]**alpha, k[j]**alpha, k[j]**alpha_H, k[j]**alpha
    y_n_L[j], y_delta_L[j], y_alpha_L[j], y_s_L[j] = k[j]**alpha, k[j]**alpha, k[j]**alpha_L, k[j]**alpha
    i_n_H[j], i_delta_H[j], i_alpha_H[j], i_s_H[j] = s*(k[j]**alpha), s*(k[j]**alpha), s*(k[j]**alpha_H), s_H*(k[j]**alpha)
    i_n_L[j], i_delta_L[j], i_alpha_L[j], i_s_L[j] = s*(k[j]**alpha), s*(k[j]**alpha), s*(k[j]**alpha_L), s_L*(k[j]**alpha)
    dep_n_H[j], dep_delta_H[j], dep_alpha_H[j], dep_s_H[j] = (n_H + delta)*k[j], (n + delta_H)*k[j], (n + delta)*k[j], (n + delta)*k[j]
    dep_n_L[j], dep_delta_L[j], dep_alpha_L[j], dep_s_L[j] = (n_L + delta)*k[j], (n + delta_L)*k[j], (n + delta)*k[j], (n + delta)*k[j]
    k_dot_n_H[j], k_dot_delta_H[j], k_dot_alpha_H[j], k_dot_s_H[j] = i_n_H[j] - dep_n_H[j], i_delta_H[j] - dep_delta_H[j], i_alpha_H[j] - dep_alpha_H[j], i_s_H[j] - dep_s_H[j]
    k_dot_n_L[j], k_dot_delta_L[j], k_dot_alpha_L[j], k_dot_s_L[j] = i_n_L[j] - dep_n_L[j], i_delta_L[j] - dep_delta_L[j], i_alpha_L[j] - dep_alpha_L[j], i_s_L[j] - dep_s_L[j]

st_n_H, st_delta_H, st_alpha_H, st_s_H = np.argmin(k_dot_n_H[1:size]**2), np.argmin(k_dot_delta_H[1:size]**2), np.argmin(k_dot_alpha_H[1:size]**2), np.argmin(k_dot_s_H[1:size]**2)
st_n_L, st_delta_L, st_alpha_L, st_s_L = np.argmin(k_dot_n_L[1:size]**2), np.argmin(k_dot_delta_L[1:size]**2), np.argmin(k_dot_alpha_L[1:size]**2), np.argmin(k_dot_s_L[1:size]**2)

fig, ax = plt.subplots()
ax.plot(k, y,'b')
ax.plot(k, i,'b')
ax.plot(k, dep,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, y_n_H,'g', label = 'n maior')
ax.plot(k, i_n_H,'g')
ax.plot(k, dep_n_H,'g')
plt.axvline(x=k[st_n_H+1], color='g', linestyle = 'dotted')
ax.plot(k, y_n_L,'r', label = 'n menor')
ax.plot(k, i_n_L,'r')
ax.plot(k, dep_n_L,'r')
plt.axvline(x=k[st_n_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
ax.set_ylabel('EC: n')
ax.set_xlabel('k')
plt.legend()
plt.show()


fig, ax = plt.subplots()
ax.plot(k, k_dot,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, k_dot_n_H,'g', label = 'n maior')
plt.axvline(x=k[st_n_H+1], color='g', linestyle = 'dotted')
ax.plot(k, k_dot_n_L,'r', label = 'n menor')
plt.axvline(x=k[st_n_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
plt.axhline(y=0, color='black')
ax.set_ylabel('EC: n')
ax.set_xlabel('k')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(k, y,'b')
ax.plot(k, i,'b')
ax.plot(k, dep,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, y_delta_H,'g', label = r'$\delta$ maior')
ax.plot(k, i_delta_H,'g')
ax.plot(k, dep_delta_H,'g')
plt.axvline(x=k[st_delta_H+1], color='g', linestyle = 'dotted')
ax.plot(k, y_delta_L,'r', label = r'$\delta$ menor')
ax.plot(k, i_delta_L,'r')
ax.plot(k, dep_delta_L,'r')
plt.axvline(x=k[st_delta_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
ax.set_ylabel(r'EC: $\delta$')
ax.set_xlabel('k')
plt.legend()
plt.show()


fig, ax = plt.subplots()
ax.plot(k, k_dot,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, k_dot_delta_H,'g', label = r'$\delta$ maior')
plt.axvline(x=k[st_delta_H+1], color='g', linestyle = 'dotted')
ax.plot(k, k_dot_delta_L,'r', label = r'$\delta$ menor')
plt.axvline(x=k[st_delta_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
plt.axhline(y=0, color='black')
ax.set_ylabel(r'EC: $\delta$')
ax.set_xlabel('k')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(k, y,'b')
ax.plot(k, i,'b')
ax.plot(k, dep,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, y_alpha_H,'g', label = r'$\alpha$ maior')
ax.plot(k, i_alpha_H,'g')
ax.plot(k, dep_alpha_H,'g')
plt.axvline(x=k[st_alpha_H+1], color='g', linestyle = 'dotted')
ax.plot(k, y_alpha_L,'r', label = r'$\alpha$ menor')
ax.plot(k, i_alpha_L,'r')
ax.plot(k, dep_alpha_L,'r')
plt.axvline(x=k[st_alpha_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
ax.set_ylabel(r'EC: $\alpha$')
ax.set_xlabel('k')
plt.legend()
plt.show()


fig, ax = plt.subplots()
ax.plot(k, k_dot,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, k_dot_alpha_H,'g', label = r'$\alpha$ maior')
plt.axvline(x=k[st_alpha_H+1], color='g', linestyle = 'dotted')
ax.plot(k, k_dot_alpha_L,'r', label = r'$\alpha$ menor')
plt.axvline(x=k[st_alpha_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
plt.axhline(y=0, color='black')
ax.set_ylabel(r'EC: $\alpha$')
ax.set_xlabel('k')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(k, y,'b')
ax.plot(k, i,'b')
ax.plot(k, dep,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, y_s_H,'g', label = 's maior')
ax.plot(k, i_s_H,'g')
ax.plot(k, dep_s_H,'g')
plt.axvline(x=k[st_s_H+1], color='g', linestyle = 'dotted')
ax.plot(k, y_s_L,'r', label = 's menor')
ax.plot(k, i_s_L,'r')
ax.plot(k, dep_s_L,'r')
plt.axvline(x=k[st_s_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
ax.set_ylabel('EC: s')
ax.set_xlabel('k')
plt.legend()
plt.show()


fig, ax = plt.subplots()
ax.plot(k, k_dot,'b')
plt.axvline(x=steady[0], color='b', linestyle = 'dotted')
ax.plot(k, k_dot_s_H,'g', label = 's maior')
plt.axvline(x=k[st_s_H+1], color='g', linestyle = 'dotted')
ax.plot(k, k_dot_s_L,'r', label = 's menor')
plt.axvline(x=k[st_s_L+1], color='r', linestyle = 'dotted')
plt.axvline(x=0, color='black', linestyle = 'dotted')
plt.axhline(y=0, color='black')
ax.set_ylabel('EC: s')
ax.set_xlabel('k')
plt.legend()
plt.show()
```

```python
"""

Transition Dynamics / Dinâmica de Transição

"""

k0, k1, k2, k3, k4, k5 = np.zeros(20), np.zeros(20), np.zeros(20), np.zeros(20), np.zeros(20), np.zeros(20)
dk0, dk1, dk2, dk3, dk4, dk5 = np.zeros(19), np.zeros(19), np.zeros(19), np.zeros(19), np.zeros(19), np.zeros(19)
k0[0], k1[0], k2[0], k3[0], k4[0], k5[0] = 0, 0.1, 1, 2, 3, 4

for t in range(1,20):
    k0[t] = k0[t-1] + s*(k0[t-1]**alpha) - (n+delta)*k0[t-1]
    k1[t] = k1[t-1] + s*(k1[t-1]**alpha) - (n+delta)*k1[t-1]
    k2[t] = k2[t-1] + s*(k2[t-1]**alpha) - (n+delta)*k2[t-1]
    k3[t] = k3[t-1] + s*(k3[t-1]**alpha) - (n+delta)*k3[t-1]
    k4[t] = k4[t-1] + s*(k4[t-1]**alpha) - (n+delta)*k4[t-1]
    k5[t] = k5[t-1] + s*(k5[t-1]**alpha) - (n+delta)*k5[t-1]

for t in range(0,19):
    dk0[t] = k0[t+1] - k0[t]
    dk1[t] = k1[t+1] - k1[t]
    dk2[t] = k2[t+1] - k2[t]
    dk3[t] = k3[t+1] - k3[t]
    dk4[t] = k4[t+1] - k4[t]
    dk5[t] = k5[t+1] - k5[t]

fig, ax = plt.subplots()
ax.plot(k0,'r', label = r'$k_0=0$' )
ax.plot(k1,'g', label = r'$k_0=0.1$' )
ax.plot(k2,'b', label = r'$k_0=1$' )
ax.plot(k3,'y', label = r'$k_0=2$' )
ax.plot(k4,'c', label = r'$k_0=3$' )
ax.plot(k5,'m', label = r'$k_0=4$' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black', linestyle = 'dotted')
ax.set_ylabel('k(t)')
ax.set_xlabel('t')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(dk0,'r', label = r'$k_0=0$' )
ax.plot(dk1,'g', label = r'$k_0=0.1$' )
ax.plot(dk2,'b', label = r'$k_0=1$' )
ax.plot(dk3,'y', label = r'$k_0=2$' )
ax.plot(dk4,'c', label = r'$k_0=3$' )
ax.plot(dk5,'m', label = r'$k_0=4$' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black', linestyle = 'dotted')
ax.set_ylabel(r'$\dot{k}$')
ax.set_xlabel('t')
plt.legend()
plt.show()
```

```python
"""

Golden Rule / Regra de Ouro

"""

s_grid = np.linspace(0,1,size)
kss, yss, css = np.zeros(size), np.zeros(size), np.zeros(size)

def ss(s):
    k = ( s / ( n + delta ) )**( 1 / ( 1 - alpha ) )
    y = ( s / ( n + delta ) )**( alpha / ( 1 - alpha ) )
    c = y - s*y
    return (k, y, c)

for i in range(0,size):
    kss[i] = ss(s_grid[i])[0]
    yss[i] = ss(s_grid[i])[1]
    css[i] = ss(s_grid[i])[2]

gr = np.argmax(css)
cgr = css[gr]
kgr = kss[gr]

fig, ax = plt.subplots()
ax.plot(s_grid, css,'b', label = r'$c_{ss}$' )
plt.axvline(x=0, color='black')
plt.axvline(x=s_grid[gr], color='b', linestyle = 'dotted')
plt.axhline(y=0, color='black')
plt.axhline(y=cgr, color='b', linestyle = 'dotted')
ax.set_ylabel('c')
ax.set_xlabel('s')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(kss, css,'b', label = r'$c_{ss}$' )
plt.axvline(x=0, color='black')
plt.axvline(x=kgr, color='b', linestyle = 'dotted')
plt.axhline(y=0, color='black')
plt.axhline(y=cgr, color='b', linestyle = 'dotted')
ax.set_ylabel('c')
ax.set_xlabel('k')
plt.legend()
plt.show()

print('Poupança de Regra de Ouro', s_grid[gr] , "Consumo de Regra de Ouro", cgr )
```

```python
"""

Variable Dynamics / Dinâmica das Variáveis

"""

L, k, y, K, Y = np.zeros(20), np.zeros(20), np.zeros(20), np.zeros(20), np.zeros(20)
dL, dk, dy, dK, dY = np.zeros(19), np.zeros(19), np.zeros(19), np.zeros(19), np.zeros(19)
L[0], k[0] = 1, 0.5
y[0] = k[0]**alpha 
K[0], Y[0] = k[0]*L[0], y[0]*L[0]

for t in range(1,20):
    L[t] = L[t-1]*(1+n) 
    k[t] = k[t-1] + s*(k[t-1]**alpha) - (n+delta)*k[t-1]
    y[t] = k[t]**alpha
    K[t] = k[t]*L[t]
    Y[t] = y[t]*L[t]

for t in range(0,19):
    dL[t] = np.log(L[t+1]) - np.log(L[t])
    dk[t] = np.log(k[t+1]) - np.log(k[t])
    dy[t] = np.log(y[t+1]) - np.log(y[t])
    dK[t] = np.log(K[t+1]) - np.log(K[t])
    dY[t] = np.log(Y[t+1]) - np.log(Y[t])

fig, ax = plt.subplots()
ax.plot(k,'g', label = 'k' )
ax.plot(y,'b', label = 'y' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black')
plt.axhline(y=steady[0], color='black', linestyle = 'dotted')
plt.axhline(y=steady[1], color='black', linestyle = 'dotted')
ax.set_ylabel('k,y')
ax.set_xlabel('t')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(L,'r', label = 'L' )
ax.plot(K,'y', label = 'K' )
ax.plot(Y,'c', label = 'Y' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black')
ax.set_ylabel('L,K,Y')
ax.set_xlabel('t')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(dk,'g', label = '\u0394 k' )
ax.plot(dy,'b', label = '\u0394 y' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black')
ax.set_ylabel('k,y')
ax.set_xlabel('t')
plt.legend()
plt.show()

fig, ax = plt.subplots()
ax.plot(dL,'r', label = '\u0394 L' )
ax.plot(dK,'y', label = '\u0394 K' )
ax.plot(dY,'c', label = '\u0394 Y' )
plt.axvline(x=0, color='black')
plt.axhline(y=0, color='black')
ax.set_ylabel('L,K,Y')
ax.set_xlabel('t')
plt.legend()
plt.show()
```
